```python
from scdesigner.datasets import pancreas

example_sce = pancreas()
```

```python
from scdesigner.simulators import PositiveNMF

formula = "~ bs(pseudotime, degree=5)"
model = PositiveNMF(formula, maxIter=500, epochs=50)
model.fit(example_sce, lr=2)
model.predict()
model.sample()
```

```python
model.predict(obs=example_sce.obs[:20])
model.sample(obs=example_sce.obs[:20])
```
